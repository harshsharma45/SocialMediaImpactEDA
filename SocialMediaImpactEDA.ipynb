```python
continuous_cols = [
    'Daily_Usage_Hours', 'Sleep_Duration_Hours', 'Perceived_Stress_Score',
    'Mental_Health_Index', 'Academic_Performance_GPA', 'Weekend_Extra_Hours'
]

fig, axes = plt.subplots(2, 3, figsize=(16, 9))
for ax, col in zip(axes.flatten(), continuous_cols):
    sns.histplot(df[col], kde=True, ax=ax, color='blue', edgecolor='black')
    ax.set_title(f'Distribution of {col}', fontsize=11, fontweight='bold')
    ax.set_xlabel(col)
    ax.set_ylabel('Student Count')

plt.tight_layout()
plt.show()
```

```python
df['Total_Weekly_Hours'] = (df['Daily_Usage_Hours'] * 5) + ((df['Daily_Usage_Hours'] + df['Weekend_Extra_Hours']) * 2)
df['Usage_Category'] = pd.cut(
    df['Daily_Usage_Hours'], 
    bins=[0, 3, 6, 9, 15], 
    labels=['Low (<3h)', 'Moderate (3-6h)', 'High (6-9h)', 'Extreme (>9h)']
)
df['Sleep_Health_Score'] = df['Sleep_Duration_Hours'] * df['Sleep_Quality_Score']
df[['Student_ID', 'Daily_Usage_Hours', 'Weekend_Extra_Hours', 'Total_Weekly_Hours', 'Usage_Category', 'Sleep_Health_Score']].head()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.barplot(data=df, x='Usage_Category', y='Mental_Health_Index', palette='crest', ax=axes[0], errorbar=None)
axes[0].set_title('Mental Health Index by Usage', fontweight='bold')
axes[0].set_xlabel('Usage')
axes[0].set_ylabel('Mean Mental Health Index')

sns.barplot(data=df, x='Usage_Category', y='Academic_Performance_GPA', palette='flare', ax=axes[1], errorbar=None)
axes[1].set_title('Academic GPA by Usage', fontweight='bold')
axes[1].set_xlabel('Usage')
axes[1].set_ylabel('Mean GPA')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 8))
numeric_cols = df.select_dtypes(include=[np.number, bool])
corr_matrix = numeric_cols.corr()

sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1, linewidths=0.5)
plt.title('Correlation Heatmap of Numerical Features', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()
```
